```python
import chunk
from langchain_text_splitters import KonlpyTextSplitter

#data/appendix-keywords.txt 파일을 열어서 읽기 모드로 연다.
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read() # 파일의 내용을 읽어서 file 변수에 저장

# KonlpyTextSplitter를 사용하여 텍스트를 분할기 객체를 생성
text_splitter = KonlpyTextSplitter(chunk_size=200, chunk_overlap=50)

texts = text_splitter.split_text(file) # 한국어 문서를 문장 단위로 분할
print(texts[0]) # 분할된 문서의 첫 번째 문서를 출력
```

```python
from langchain_text_splitters import CharacterTextSplitter
from transformers import GPT2TokenizerFast

# GPT2TokenizerFast를 사용하여 토크나이저 객체를 생성
hf_tokenizer = GPT2TokenizerFast.from_pretrained("gpt2")

#data/appendix-keywords.txt 파일을 열어서 읽기 모드로 연다.
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read() # 파일의 내용을 읽어서 file 변수에 저장

text_splitter = CharacterTextSplitter.from_huggingface_tokenizer(
    # Hugging Face 토크나이저를 사용하여 CharacterTextSplitter 객체를 생성
    hf_tokenizer,
    chunk_size=300, # 청크 크기 설정
    chunk_overlap=50 # 청크 간의 중복되는 문자 수 설정
)

#state_of_the_union 텍스트를 분할하여 texts 변수에 저장
texts = text_splitter.split_text(file)
print(texts[1]) # 분할된 문서의 첫 번째 문서를 출력

```

### 의미 단위로 분할하기

```python
with open ("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read() # 파일의 내용을 읽어서 file 변수에 저장

print(file[:350]) # 파일 내용 일부 출력
```

```python
from langchain_experimental.text_splitter import SemanticChunker
from langchain_openai.embeddings import OpenAIEmbeddings
from dotenv import load_dotenv

load_dotenv() # .env 파일의 환경 변수를 로드

text_splitter = SemanticChunker(OpenAIEmbeddings()) # semanticchunker 초기화
```

```python
with open("./data/appendix-keywords.txt", encoding="utf-8") as f:
    file = f.read()

chunks = text_splitter.split_text(file) # 텍스트를 의미론적 청크로 분할
print(chunks[0]) # 분할된 문서의 첫 번째 청크를 출력
```

```python
docs = text_splitter.create_documents([file]) # 텍스트를 의미론적 청크로 분할하여 문서 객체 생성
print(docs[0]) # 분할된 문서의 첫 번째 문서를 출력
```

```python
text_splitter = SemanticChunker(
    OpenAIEmbeddings(), # SemanticChunker를 초기화할 때 OpenAIEmbeddings를 사용하여 의미론적 임베딩을 생성
    breakpoint_threshold_type="percentile",
    breakpoint_threshold_amount=70,
)

docs = text_splitter.create_documents([file]) # 텍스트를 의미론적 청크로 분할하여 문서 객체 생성
for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc.page_content) # 분할된 문서의 내용을 출력
    print("===" * 20) # 구분선 출력

print(len(docs)) # 분할된 문서의 총 개수를 출력
```

```python
text_splitter = SemanticChunker(
    OpenAIEmbeddings(), # semanticchunker 초기화
    breakpoint_threshold_type="standard_deviation", # 표준편차 기준
    breakpoint_threshold_amount=1.25, # 표준편차 기준으로 청크를 나누는 임계값 설정
)

docs = text_splitter.create_documents([file]) # 텍스트를 의미론적 청크로 분할하여 문서 객체 생성
for i, doc in enumerate(docs[:5]):
    print(f"[Chunk {i}]", end="\n\n")
    print(doc.page_content) # 분할된 문서의 내용을 출력
    print("===" * 20) # 구분선 출력

print(len(docs)) # 분할된 문서의 총 개수를 출력
```
