### WebBaseLoader

```python
from dotenv import load_dotenv

load_dotenv()
```

```python
import bs4
from langchain_community.document_loaders import WebBaseLoader

loader = WebBaseLoader( # 뉴스 기사 내용을 로드
    web_paths=("https://n.news.naver.com/article/437/0000378416"),
)
```

```python
loader = WebBaseLoader( # 뉴스 기사 내용을 로드
    web_paths=("https://n.news.naver.com/article/437/0000378416",),
    bs_kwargs=dict(
        parse_only=bs4.SoupStrainer(
            "div",
                attrs={"class": ["newsct_article _article_body", "media_end_head_title"]}, # 기사 본문과 제목만 가져오기
        )
    ),
    header_template={ 
        "User_Agent": "Mozilla/5.0 (windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/102.0.0.0 Safari/537.36", # 브라우저 식별자
    },
)
```

```python
docs = loader.load()
print(f"문서의 수: {len(docs)}")
docs
```

```python
loader.requests_kwargs = {"verify": True} # SSL 인증 우회
docs = loader.load() # 데이터 로드
```

```python
import nest_asyncio # 비동기 작업을 위한 라이브러리 jupyter notebook 에서만 실행

nest_asyncio.apply() 
loader.requests_per_second = 1 # 초당 요청 수 설정
docs = loader.load() # 비동기 로드

docs # 결과 출력
```
