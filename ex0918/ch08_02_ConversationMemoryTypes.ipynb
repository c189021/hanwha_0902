```python
from langchain_classic.memory import ConversationBufferWindowMemory

memory = ConversationBufferWindowMemory(k=2, return_messages=True)

memory.save_context(
    inputs={"human": "안녕하세요. 오늘 날씨가 어떤가요?"},
    outputs={"ai": "안녕하세요! 오늘 날씨는 맑고 화창합니다."},
)

memory.save_context(
    inputs={"human": "좋아요. 내일은 어떨까요?"},
    outputs={"ai": "내일은 흐리고 비가 올 가능성이 있습니다."},
)

memory.save_context(
    inputs={"human": "그럼 모레는요?"},
    outputs={"ai": "모레는 맑고 따뜻한 날씨가 예상됩니다."},
)

memory.save_context(
    inputs={"human": "좋아요. 그럼 이번 주말 날씨는 어떨까요?"},
    outputs={"ai": "이번 주말은 대체로 맑고 기온도 적당할 것으로 예상됩니다."},
)

memory.save_context(
    inputs={"human": "감사합니다. 그럼 다음 주 날씨는 어떨까요?"},
    outputs={"ai": "다음 주는 대체로 흐리고 비가 올 가능성이 있습니다. 날씨 예보를 참고하시기 바랍니다."},
)

memory.save_context(
    inputs={"human": "좋아요. 그럼 다음 달 날씨는 어떨까요?"},
    outputs={"ai": "다음 달은 대체로 맑고 기온도 적당할 것으로 예상됩니다. 하지만 날씨는 변동성이 있으므로 주기적으로 확인하시기 바랍니다."},
)
```

```python
memory.load_memory_variables({})["history"]  # 메모리 로드
```

### 대화 토큰 버퍼 메모리

```python
from langchain_classic.memory import ConversationTokenBufferMemory
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv

load_dotenv()


llm = ChatOpenAI(model_name="gpt-4o") # LLM 모델 생성

memory = ConversationTokenBufferMemory( # 메모리 설정
    llm=llm,
    max_token_limit=200, # 최대토큰길이제한(이게 작으면 몇개 저장 안 됨)
    return_messages=True
)
```

### 대화 엔티티 메모리

```python
from langchain_openai import ChatOpenAI
from langchain_classic.chains import ConversationChain
from langchain_classic.memory import ConversationEntityMemory
from langchain_classic.memory.prompt import ENTITY_MEMORY_CONVERSATION_TEMPLATE
from dotenv import load_dotenv

load_dotenv()

print(ENTITY_MEMORY_CONVERSATION_TEMPLATE.template)  # 엔티티 메모리 프롬프트 확인
```

```python
llm = ChatOpenAI(model_name="gpt-4o", temperature=0) # LLM 모델 생성

conversation = ConversationChain(
    llm=llm,
    prompt=ENTITY_MEMORY_CONVERSATION_TEMPLATE,
    memory=ConversationEntityMemory(llm=llm),
)
```

```python
conversation.predict(
    input="테디와 셜리는 한 회사에서 일하는 동료입니다."
    "테디는 개발자이고 셜리는 디자이너입니다."
    "그들은 최근 회사에서 일하는 것을 그만두고 자신들의 회사를 차릴 계획을 세우고 있습니다."
)
```

```python
conversation.memory.entity_store.store # 엔티티 메모리 확인
```

### 대화 지식 그래프 메모리

```python
from langchain_openai import ChatOpenAI
from langchain_classic.memory import ConversationKGMemory
from dotenv import load_dotenv

load_dotenv()

llm = ChatOpenAI(model_name="gpt-4o", temperature=0) # LLM 모델 생성
```

```python
memory = ConversationKGMemory(llm=llm, return_messages=True) # KG 메모리 생성
memory.save_context(
    {"input": "이쪽은 Pangyo에 거주중인 김셜리씨 입니다."},
    {"output": "김셜리씨는 누구시죠?"},
)
memory.save_context(
    {"input": "김셜리씨는 우리 회사의 신입 디자이너입니다."},
    {"output": "만나서 반갑습니다."},
)

memory.load_memory_variables({"input": "김셜리씨는 누구입니까?"})  # 메모리 로드
```

```python
from langchain_classic.prompts.prompt import PromptTemplate
from langchain_classic.chains import ConversationChain

llm = ChatOpenAI(temperature=0)

template = """The following is a friendly conversation"...
The AI ONLY uses information containded in the "Relevant Information" section ...

Relevant Information:

{history}

Conversation:
Human: {input}
AI:"""

prompt = PromptTemplate(
    input_variables=["history", "input"], template=template)

conversation_with_kg = ConversationChain(
    llm=llm,
    prompt=prompt,

    memory=ConversationKGMemory(llm=llm)
)
```

```python
conversation_with_kg.predict(
    input="My name is Teddy. Shirley is a coworker of mine, and she's a new designer at our company."
)
```

```python
conversation_with_kg.memory.load_memory_variables({"input": "who is Shirley?"})  # 메모리 로드
```
