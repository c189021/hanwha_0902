```python
from langchain_classic.memory import ConversationBufferMemory

memory = ConversationBufferMemory()
memory
```

```python
memory.save_context(
    inputs={
        "human": "안녕하세요, 비대면으로 은행 계좌를 개설하고 싶습니다. 어떻게 시작해야 하나요?"
    },
    outputs={
        "ai": "안녕하세요, 계좌 개설을 원하시다니 기쁩니다. 먼저, 본인 인증을 위해 신분증을 준비해 주시겠어요?"
    },
)
```

```python
memory.save_context(
    inputs={"human": "신분증을 준비했습니다. 다음 단계는 무엇인가요?"},
    outputs={
        "ai": "좋습니다. 다음 단계로는 계좌 개설 신청서를 작성해야 합니다. 온라인으로 신청서를 작성하시겠습니까, 아니면 지점에 방문하시겠습니까?"
    },
)
memory.save_context(
    inputs={"human": "사진을 업로드했습니다. 본인 인증은 어떻게 진행되나요?"},
    outputs={
        "ai": "사진을 확인했습니다. 본인 인증은 업로드하신 신분증과 사진을 비교하여 진행됩니다. 잠시만 기다려 주세요."
    },
)
memory.save_context(
    inputs={"human": "본인 인증이 완료되었습니다. 계좌 개설이 진행되나요?"},
    outputs={
        "ai": "본인 인증이 완료되었습니다. 이제 계좌 개설 절차를 진행하겠습니다. 잠시만 기다려 주세요."
    },
)
```

```python
print(memory.load_memory_variables({})["history"])
```

```python
memory.save_context(
    inputs={"human": "정보를 모두 입력했습니다. 다음 단계는 무엇인가요?"},
    outputs={
        "ai": "입력해 주신 정보를 확인했습니다. 계좌 개설 절차가 거의 끝났습니다. 마지막으로 이용 약관에 동의해 주시고, 계좌 개설을 최종 확인해 주세요"
    },
    
)
memory.save_context(
    inputs={"human": "이용 약관에 동의했습니다. 계좌 개설을 최종 확인합니다."},
    outputs={
        "ai": "이용 약관에 동의해 주셔서 감사합니다. 계좌 개설이 완료되었습니다. 축하드립니다! 이제부터 계좌를 이용하실 수 있습니다."
    },
)
```

```python
print(memory.load_memory_variables({})["history"])
```

```python
memory = ConversationBufferMemory(return_messages=True)

memory.save_context(
    inputs={
        "human": "안녕하세요, 비대면으로 은행 계좌를 개설하고 싶습니다. 어떻게 시작해야 하나요?"
    },
    outputs={
        "ai": "안녕하세요! 비대면으로 은행 계좌를 개설해 드릴 수 있습니다. 먼저 본인 인증을 진행해 주세요."
    },
)

memory.save_context(
    inputs={"human": "신분증을 준비했습니다. 다음 단계는 무엇인가요?"},
    outputs={
        "ai": "감사합니다. 신분증 앞뒤 촬영해서 올려주세요."
    },
)

memory.save_context(
    inputs={"human": "사진을 업로드했습니다. 본인 인증은 어떻게 진행되나요?"},
    outputs={
        "ai": "사진을 확인했습니다. 본인 인증은 업로드하신 신분증과 사진을 비교하여 진행됩니다. 잠시만 기다려 주세요."
    },
)
```
