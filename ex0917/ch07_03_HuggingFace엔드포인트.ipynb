```python
import os
from dotenv import load_dotenv
from langchain_teddynote import logging
from huggingface_hub import login

load_dotenv()  # .env 파일에서 환경 변수를 로드
logging.langsmith("CH04-Models")

login(os.environ.get("HUGGINGFACEHUB_API_TOKEN"))  # Hugging Face Hub에 로그인
```

```python
from langchain_core.prompts import PromptTemplate

template = """Question: {question}

Answer:"""

template = """<|system|>
Your are a helpful assistant.s<|end|>
<|user|>
{question}<|end|>
<|assistant|>"""

prompt = PromptTemplate.from_template(template)
```

```python
import os
from langchain_core.output_parsers import StrOutputParser
from langchain_huggingface import ChatHuggingFace, HuggingFaceEndpoint

repo_id = "openai/gpt-oss-120b"

llm = HuggingFaceEndpoint(
    provider="auto", # 사용 가능한 provider 중 자동 선택
    repo_id=repo_id, # Hugging Face Hub에서 사용할 모델의 ID
    max_new_tokens=256, # 생성할 최대 토큰 길이
    temperature=0.1, # 생성되는 텍스트의 다양성을 제어하는 하이퍼파라미터
    huggingfacehub_api_token=os.environ.get("HUGGINGFACEHUB_API_TOKEN"), # Hugging Face Hub API 토큰
)

chat_llm = ChatHuggingFace(llm=llm)

chain = prompt | chat_llm | StrOutputParser() # 프롬프트와 LLM을 연결하여 체인을 생성
response = chain.invoke({"question": "What is the capital of South korea?"}) # 체인을 실행하여 질문에 대한 답변을 생성

print(response) # 생성된 답변을 출력
```
